```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as pyplot
```

```python
# 1. load dataset stunting

df = pd.read_csv('data/stunting_nasional.csv')
print('Rows:', len(df))
print('Columns:', df.shape[1])
print(df.info())
```

```python
# 2. check missing values and Memory usage

print("Total Memori:", df.memory_usage(deep=True).sum() / (1024**2), 'MB')
print(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt
# 3. Simple aggregation example (Disesuaikan dengan kolom stunting: Usia_Bulan & Z_Score_TB_U)
if 'Usia_Bulan' in df.columns and 'Z_Score_TB_U' in df.columns:
    summary_usia = df.groupby('Usia_Bulan')['Z_Score_TB_U'].mean().reset_index()
    plt.plot(summary_usia['Usia_Bulan'], summary_usia['Z_Score_TB_U'], marker='o', color='teal')
    plt.title('Rata-rata Z-Score TB/U Berdasarkan Usia (Bulan)')
    plt.xlabel('Usia (Bulan)')
    plt.ylabel('Rata-rata Z-Score')
    plt.grid(True)
    plt.show()
```

```python
# 4. Save cleaned sample (1% sample sesuai instruksi modul)
df_sample = df.sample(frac=0.01, random_state=42)
df_sample.to_csv('data/stunting_sample.csv', index=False)
print("Sampel bersih berhasil diekspor!")
```
